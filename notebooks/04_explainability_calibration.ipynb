```python
DATA_PATH = PROJECT_ROOT / "data" / "telecom_churn_data.csv"

df = pd.read_csv(DATA_PATH)

print("Shape:", df.shape)
display(df.head())
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nTarget-related columns:")
print([c for c in df.columns if "voice_active" in c])

print("\nMissing values:")
display(df[["voice_active_8", "voice_active_9"]].isna().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df["newly_inactive"] = (
    (df["voice_active_8"] == 1) &
    (df["voice_active_9"] == 0)
).astype(int)

print("Target distribution:")
display(df["newly_inactive"].value_counts())

print("\nTarget proportion:")
display(df["newly_inactive"].value_counts(normalize=True))
```
